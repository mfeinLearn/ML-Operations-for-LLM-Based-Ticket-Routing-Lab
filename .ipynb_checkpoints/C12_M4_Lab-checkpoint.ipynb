## Step 0: Environment Setup and Data Ingestion

```python
import os
# This env's TensorFlow was built for protobuf 3.x and crashes on protobuf 6.
# The lab uses PyTorch through sentence-transformers, so skip TensorFlow.
os.environ["USE_TF"] = "0"

%pip install pandas scikit-learn sentence-transformers mlflow matplotlib seaborn

import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score, f1_score, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import mlflow
import datetime
import json
```

## Step 1: Data Intake & Feature Management

```python
# Replace this with your custom dataset function
# def create_ticket_dataset():
#     return pd.DataFrame({
#         'text': ["I can't log in", "How do I change billing info?", "App crashes on launch", "Need to cancel my subscription"],
#         'label': ["Technical Support", "Billing", "Technical Support", "Account Management"]
#     })
def create_ticket_dataset():
    data = [
        # Technical Support (clean)
        ("I can't log in to my account", "Technical Support"),
        ("App crashes every time I open it", "Technical Support"),
        ("Getting a 500 error when I try to upload", "Technical Support"),
        ("Password reset link is not working", "Technical Support"),
        ("The app freezes on the home screen", "Technical Support"),
        ("Can't connect to the server, timeout error", "Technical Support"),
        ("Login button does nothing when clicked", "Technical Support"),
        ("App keeps crashing on Android", "Technical Support"),
        # Billing
        ("How do I change my billing information?", "Billing"),
        ("I was charged twice this month", "Billing"),
        ("Need to update credit card on file", "Billing"),
        ("Invoice is wrong, overcharged", "Billing"),
        ("How do I download my receipt?", "Billing"),
        ("Payment failed, card declined", "Billing"),
        ("Change subscription plan pricing", "Billing"),
        ("Refund request for last month", "Billing"),
        # Account Management
        ("Need to cancel my subscription", "Account Management"),
        ("How do I delete my account?", "Account Management"),
        ("Change email address on my profile", "Account Management"),
        ("Want to upgrade my plan", "Account Management"),
        ("Close my account permanently", "Account Management"),
        ("Update my username", "Account Management"),
        ("Pause my subscription for 3 months", "Account Management"),
        ("Transfer account to another email", "Account Management"),
        # Noisy / ambiguous
        ("help me pls the thing is broken and money is weird", "Technical Support"),
        ("billing login problem cancel?", "Billing"),
        ("account crashed and charged me", "Account Management"),
    ]
    return pd.DataFrame(data, columns=["text", "label"])

model_name = 'all-MiniLM-L6-v2'
embedder = SentenceTransformer(model_name)

raw_df = create_ticket_dataset()
X = embedder.encode(raw_df['text'].tolist())
y = raw_df['label']

X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)
```

## Step 2: Experimentation & Training

```python
mlflow.set_tracking_uri("file:mlruns")
mlflow.set_experiment("ticket_routing_llm")

models = {
    "LogisticRegression": LogisticRegression(max_iter=500),
    "RandomForest": RandomForestClassifier(n_estimators=100, random_state=42),
    "SVM (Linear Kernel)": SVC(kernel='linear', probability=True, random_state=42)
}

for name, model in models.items():
    with mlflow.start_run(run_name=name):
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        f1 = f1_score(y_test, y_pred, average='weighted')

        mlflow.log_param("model_type", name)
        mlflow.log_param("embedding_model", model_name)
        mlflow.log_metric("accuracy", acc)
        mlflow.log_metric("f1_score", f1)

        print(f"\nModel: {name}")
        print(classification_report(y_test, y_pred))

        ConfusionMatrixDisplay.from_predictions(y_test, y_pred, xticks_rotation=45)
        plt.title(f"Confusion Matrix: {name}")
        plt.tight_layout()
        plt.show()
```

## Step 3: Automated Validation (CI)

```python
def test_embedding_output(X):
    assert X.shape[1] == 384, "Embedding dimensionality mismatch."
    print("Test passed: Embedding output shape correct.")

def test_class_distribution(y):
    counts = pd.Series(y).value_counts()
    assert all(counts > 1), "Class imbalance too severe."
    print("Test passed: Class distribution sufficient.")

print("Running validation tests...")
test_embedding_output(X_train)
test_class_distribution(y_train)
```
